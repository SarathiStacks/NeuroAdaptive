```python
import os
from pathlib import Path
from PIL import Image
import numpy as np
import time

RAW_DIR = Path(r"D:\Micro Imagine cup\NeuroAdaptive\data\ferac\raw")
OUT_DIR = Path(r"D:\Micro Imagine cup\NeuroAdaptive\data\ferac\processed")
IMG_SIZE = (224, 224)

# Create processed folders
for split in ["train", "test"]:
    for label in ["anger", "fear", "joy", "natural"]:
        (OUT_DIR / split / label).mkdir(parents=True, exist_ok=True)

def preprocess(img_path, save_path):
    img = Image.open(img_path).convert("RGB")
    img = img.resize(IMG_SIZE)
    arr = np.array(img) / 255.0  # normalize 0–1
    img_out = Image.fromarray((arr * 255).astype(np.uint8))  # back to image
    img_out.save(save_path)

start = time.time()

# Process train/test
for split in ["train", "test"]:
    for label in ["anger", "fear", "joy", "natural"]:
        folder = RAW_DIR / split / label
        for img in folder.glob("*"):
            if img.suffix.lower() in [".png", ".jpg", ".jpeg"]:
                save_path = OUT_DIR / split / label / img.name
                try:
                    preprocess(img, save_path)
                except Exception as e:
                    print(f"⚠ Skipped {img.name} ({label}) → {e}")

end = time.time()
print(f"\nFinished FERAC preprocessing in {round(end-start,2)} sec")
```
